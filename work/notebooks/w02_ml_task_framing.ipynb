## 2. Target or proxy

*What would you predict? Where does that label come from — observed outcome or a defined rule?*

The model will predict a refresh opportunity score for each webpage. Since no direct label exists, a proxy target can be created using observed engagement signals such as impressions, clicks, CTR, sessions, and pageviews. Higher engagement after refresh indicates better content opportunities.

```python
# This cell is for CODE (numbers, a query, a check).
# Write your text answer in the cell ABOVE this one — typing sentences here breaks Run All.
target = {
    "Prediction": "Refresh Opportunity Score",
    "Type": "Proxy Target",
    "Signals": [
        "impressions_90d",
        "clicks_90d",
        "pageviews_90d",
        "sessions_90d",
        "users_90d"
    ]
}

for k, v in target.items():
    print(k, ":", v)
```

## 3. Success metric

*One metric you can defend. What number means 'good'?*

The primary success metric is Precision@50 because editors review only a limited number of pages. A good model should place the highest-value refresh opportunities within the top-ranked pages.

```python
# This cell is for CODE (numbers, a query, a check).
# Write your text answer in the cell ABOVE this one — typing sentences here breaks Run All.
success_metric = "Precision@50"

baseline = 0.24
decision_tree = 0.54
random_forest = 0.74

print("Metric:", success_metric)
print("Baseline:", baseline)
print("Decision Tree:", decision_tree)
print("Random Forest:", random_forest)
```
